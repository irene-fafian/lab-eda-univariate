### Part 1: Understanding Product Categories

**Business Question**: What are the most popular product categories on Amazon UK, and how do they compare in terms of listing frequency?

1. **Frequency Tables**:
    - Generate a frequency table for the product `category`.
    - Which are the top 5 most listed product categories?

2. **Visualizations**:
    - Display the distribution of products across different categories using a bar chart. *If you face problems understanding the chart, do it for a subset of top categories.*
    - For a subset of top categories, visualize their proportions using a pie chart. Does any category dominate the listings?

---

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%matplotlib inline
```

```python
df = pd.read_csv("amz_uk_processed_data.csv")
```

```python
df.head()
```

```python
df = df.rename(columns={'categoryName': 'category'})
```

```python
# Generate a frequency table for the product category.
frequency_table = df['category'].value_counts()

proportion_table = df['category'].value_counts(normalize=True)

frequency_table, proportion_table
```

```python
# Which are the top 5 most listed product categories?
top5=df.category.value_counts().sort_values(ascending=False).head(5)
top5
```

```python
# Display the distribution of products across different categories using a bar chart. If you face problems understanding the chart, do it for a subset of top categories.

ax=sns.barplot(y=top5.index, x=top5.values, palette="Set3", hue=top5.index, legend=True )
sns.move_legend(ax, "lower right")
```


### Part 2: Delving into Product Pricing

**Business Question**: How are products priced on Amazon UK, and are there specific price points or ranges that are more common?

1. **Measures of Centrality**:
    - Calculate the mean, median, and mode for the `price` of products.
    - What's the average price point of products listed? How does this compare with the most common price point (mode)?

2. **Measures of Dispersion**:
    - Determine the variance, standard deviation, range, and interquartile range for product `price`.
    - How varied are the product prices? Are there any indicators of a significant spread in prices?

3. **Visualizations**:
    - Is there a specific price range where most products fall? Plot a histogram to visualize the distribution of product prices. *If its hard to read these diagrams, think why this is, and explain how it could be solved.*.
    - Are there products that are priced significantly higher than the rest? Use a box plot to showcase the spread and potential outliers in product pricing. 

---

```python
# Calculate the mean, median, and mode for the price of products.

mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

mean_price, median_price, mode_price
```

What's the average price point of products listed? How does this compare with the most common price point (mode)?

The average (mean) product price is $94.26, while the most common price (mode) is $9.99.
This large difference suggests the price distribution is right-skewed: most products are relatively inexpensive, but a smaller number of high-priced products raise the average substantially. The median price of $19.90 supports this interpretation, as it is much closer to the mode than to the mean.

```python
# Determine the variance, standard deviation, range, and interquartile range for product price.
variance_price = df['price'].var()
std_dev_price = df['price'].std()
min_price = df['price'].min()
max_price = df['price'].max()
range_price = max_price - min_price
quantiles_price = df['price'].quantile([0.25, 0.5, 0.75])

variance_price, std_dev_price, min_price, max_price, range_price, quantiles_price
```

How varied are the product prices? Are there any indicators of a significant spread in prices?

Product prices show a very large spread.
Prices range from $0 to $100,000, giving a range of $100,000.
The standard deviation ($360.62) is much larger than the median price ($19.90), indicating substantial variability.
The middle 50% of prices lie between $10.00 and $47.71, showing that most products are relatively low-priced.
The extremely high maximum price suggests the presence of outliers, which greatly increase the overall spread of prices.

```python
# Is there a specific price range where most products fall? Plot a histogram to visualize the distribution of product prices. If its hard to read these diagrams, think why this is, and explain how it could be solved.
fig, axes = plt.subplots()
sns.histplot(df['price'], kde=True, bins=30, color="blue", ax=axes);
plt.show()
```

The histogram shows that most products fall within the lower price ranges, with the majority priced below $50. The distribution is strongly right-skewed, meaning a small number of very expensive products pull the distribution to the right. This makes the histogram difficult to read because prices extend up to $100,000, compressing the bulk of the data near the origin. To improve readability restricting the plot to a lower price range (e.g., below the 75th percentile) could be used.

```python
# As we saw with the quartiles, most products fall in the under 50$ category so we're going to try to exclude outlier values in price by using the 75 quartile:

q75 = df['price'].quantile(0.75)

sns.histplot(df.loc[df['price'] <= q75, 'price'],bins=30,kde=True)
plt.show()
```

```python
# Are there products that are priced significantly higher than the rest? Use a box plot to showcase the spread and potential outliers in product pricing.

sns.boxplot(x = df['price'], color="blue");
```

The box plot shows the presence of significant outliers in pricing. While the middle 50% of products are priced between $10.00 and $47.71, some products have prices far above this range, reaching as high as $100,000. These extreme values create a large spread and explain why the mean price ($94.26) is much higher than the median ($19.90).


### Part 3: Unpacking Product Ratings

**Business Question**: How do customers rate products on Amazon UK, and are there any patterns or tendencies in the ratings?

1. **Measures of Centrality**:
    - Calculate the mean, median, and mode for the `rating` of products.
    - How do customers generally rate products? Is there a common trend?

2. **Measures of Dispersion**:
    - Determine the variance, standard deviation, and interquartile range for product `rating`.
    - Are the ratings consistent, or is there a wide variation in customer feedback?

3. **Shape of the Distribution**:
    - Calculate the skewness and kurtosis for the `rating` column. 
    - Are the ratings normally distributed, or do they lean towards higher or lower values?

4. **Visualizations**:
    - Plot a histogram to visualize the distribution of product ratings. Is there a specific rating that is more common?

---


```python
df = df.rename(columns={'stars': 'rating'})
```

```python
# Calculate the mean, median, and mode for the rating of products.

mean_rating = df['rating'].mean()
median_rating = df['rating'].median()
mode_rating = df['rating'].mode()[0]

mean_rating, median_rating, mode_rating
```

How do customers generally rate products? Is there a common trend?

Customers generally give low ratings, with and average rating of 2.03 put of 5. Both the median and the mode are 0.0 indicating that at least the products have a rating of 0 and that 0 is the most common rating. This is probably because there are many products with no rating.

```python
# Determine the variance, standard deviation, and interquartile range for product rating.

variance_rating = df['rating'].var()
std_dev_rating = df['rating'].std()
min_rating = df['rating'].min()
max_rating = df['rating'].max()
range_rating = max_rating - min_rating
quantiles_rating = df['rating'].quantile([0.25, 0.5, 0.75])

variance_rating, std_dev_rating, min_rating, max_rating, range_rating, quantiles_rating
```

Are the ratings consistent, or is there a wide variation in customer feedback?

The ratings show considerable variation:
- they span from 0 to 5
- standard deviation of 2.19 indicates
- the 25q and median are both 0 while the q75 is 4.4 meaning a majority of products have very low or no ratings at all while others receive very high ratings

In conclusion, no, the ratings are not consistent and appear spread out.

```python
# Calculate the skewness and kurtosis for the rating column.

skewness_rating = df['rating'].skew()
kurtosis_rating = df['rating'].kurtosis()

skewness_rating, kurtosis_rating
```

Are the ratings normally distributed, or do they lean towards higher or lower values?

No, the ratings are not normally distributed, but they don't show a very strong tendency either way.
The skewness of 0.19 despite being positive is a very low value indicating a very mild tendency towards higher ratings so the distribution of ratings is mildly right-skewed.
The kurtosis of -1.90 is less than 3 (and negative in excess kurtosis terms), indicating that the rating distribution is flatter than a normal distribution. This suggests that ratings are more evenly spread across the scale and that there are fewer extreme values than would be expected in a normally distributed dataset.

```python
# Plot a histogram to visualize the distribution of product ratings. Is there a specific rating that is more common?

fig, axes = plt.subplots()
sns.histplot(df['rating'], kde=True, bins=30, color="blue", ax=axes);
plt.show()
```

The histogram shows a very large concentration of products with a rating of 0, making 0.0 the most common rating by a significant margin. This is consistent with the earlier finding that both the median and mode are 0.0.
There is also a noticeable cluster of products with ratings between 4.0 and 5.0, indicating that many rated products receive relatively high scores. However, the large number of zero ratings dominates the distribution.
